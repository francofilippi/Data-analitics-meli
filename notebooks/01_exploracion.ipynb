## 2. Filtrado booleano

La forma de "hacer queries" en pandas. `ventas[condicion]` devuelve solo las filas donde la condicion es True.

```python
# Solo ventas pagadas del cliente tech_outlet
mask = (ventas['cliente_ml'] == 'tech_outlet') & (ventas['estado'] == 'pagado')
ventas[mask].head()
```

## 3. Agrupar y resumir (`groupby`)

El concepto mas importante del analisis de datos: **partir-aplicar-combinar**. Agrupas por una columna, aplicas una funcion (suma, promedio, conteo) y obtenes una tabla resumen.

```python
# Ingreso total por cliente
pagadas = ventas[ventas['estado'] == 'pagado']
pagadas.groupby('cliente_ml')['ingreso'].sum().sort_values(ascending=False)
```

```python
# Varias metricas a la vez por categoria
pagadas.groupby('categoria').agg(
    ingreso=('ingreso', 'sum'),
    ordenes=('order_id', 'count'),
    ticket=('ingreso', 'mean'),
).sort_values('ingreso', ascending=False)
```

## 4. Series temporales

Cuando tenes una columna de fecha, podes agrupar por dia/semana/mes con `resample`. Aca calculamos el ingreso mensual: asi se ve la **estacionalidad** (el pico de noviembre por Black Friday).

```python
mensual = pagadas.resample('ME', on='fecha')['ingreso'].sum()
mensual.plot(kind='bar', figsize=(11, 4), title='Ingreso mensual (todos los clientes)');
```
